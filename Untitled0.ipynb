```python

import pandas as pd

# Load the downloaded dataset
df = pd.read_csv("titanic.csv")

# Show the first 5 rows
df.head()
```

```python

# Check the structure of the dataset
df.info()

# Check how many values are missing in each column
df.isnull().sum()
```

```python

# 1. Fill missing Age with the median age
df['Age'].fillna(df['Age'].median(), inplace=True)

# 2. Fill missing Embarked with the most common value (mode)
df['Embarked'].fillna(df['Embarked'].mode()[0], inplace=True)

# 3. Drop the Cabin column (too many missing values)
df.drop(columns=['Cabin'], inplace=True)

# 4. Check again for missing values
df.isnull().sum()
```

```python

df.columns
```

```python

# 1. Fill missing 'age' with the median age
df['age'].fillna(df['age'].median(), inplace=True)

# 2. If 'embarked' column is missing too, let's check and handle it
# But your dataset doesn't have 'embarked' or 'cabin' — so we skip those

# 3. Recheck for missing values
df.isnull().sum()
```

```python
# Convert 'sex' column to numbers: female = 0, male = 1
df['sex'] = df['sex'].map({'female': 0, 'male': 1})

# Check if it worked
df.head()
```

```python

from sklearn.preprocessing import StandardScaler

# Create the scaler
scaler = StandardScaler()

# Apply it to 'age' and 'fare'
df[['age', 'fare']] = scaler.fit_transform(df[['age', 'fare']])

# Show updated data
df.head()
```

```python

# Step 1: Calculate IQR (Interquartile Range) for age and fare
Q1 = df[['age', 'fare']].quantile(0.25)
Q3 = df[['age', 'fare']].quantile(0.75)
IQR = Q3 - Q1

# Step 2: Remove rows where age or fare is an outlier
df = df[~((df[['age', 'fare']] < (Q1 - 1.5 * IQR)) | (df[['age', 'fare']] > (Q3 + 1.5 * IQR))).any(axis=1)]

# Step 3: Check the new shape (how many rows left)
df.shape
```

```python
df.to_csv("cleaned_titanic.csv", index=False)
```
